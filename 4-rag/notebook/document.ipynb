## Data Ingestion

```python
from langchain_core.documents import Document
```

```python
doc = Document(page_content="This is a sample document that is used to create a rag application",
    metadata={"source": "sample.txt", 
    "description": "A sample document for RAG application",
    "page": 1,
    "author": "Kumar Krishna",
    "created_at": "2024-06-05"
    } 
)
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/python_intro.txt")
document= loader.load()
print(document)
```

```python
from langchain_community.document_loaders import DirectoryLoader

directoryloader = DirectoryLoader("../data", glob="*.txt", loader_cls=TextLoader, loader_kwargs={"encoding": "utf-8"}, show_progress=False)
documents = directoryloader.load()
print(documents)
```
