### integrating with openai, google and groq

```python
import os
from dotenv import load_dotenv

os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.environ.get("GROQ_API_KEY")
```

```python
# open ai chat model

from langchain.chat_models import init_chat_model
model = init_chat_model(model="gpt-4.1", temperature=0.5, max_tokens=1000)
model
```

```python
response = model.invoke("How are you doing today?")
response.content[-1].text
```

```python
google_model = init_chat_model(model="google_genai:gemini-3.8-flash", temperature=0.5, max_tokens=1000)
google_model.invoke("How are you doing today?").content

```

```python
google_model = init_chat_model(model="google_genai:gemini-3.8-flash", temperature=0.5, max_tokens=1000)

for chunk in google_model.stream("write a 200 word paragraph on the importance of AI in healthcare"):
    print(chunk.content, end="........", flush =True)


batch_response = google_model.batch(["what is the capital of France?", "what is the capital of Germany?"],
                                    config={"max_concurrency":5})
for response in batch_response:
    print(response.content)
```

```python
groq_model = init_chat_model(model="llama-3.3-70b-versatile", 
    model_provider="groq", 
    temperature=0.5, max_tokens=1000) 
groq_model.invoke("How are you doing today?").content
```
