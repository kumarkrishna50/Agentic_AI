```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
from langchain.chat_models import init_chat_model

messages= [HumanMessage(content="How to create a simple web server in Python?"), 
 SystemMessage(content="This are a helpful coding assistant")]

import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY")

model = init_chat_model(model="google_genai:gemini-3.8-flash", temperature=0.7)
reponse = model.invoke(messages)
reponse
 
```
