### Middleware

```python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver

config = {"configurable":{"thread_id":"test-thread-id"}}

agent = create_agent(
    model="google_genai:gemini-3.8-flash",
    middleware=[SummarizationMiddleware(model="google_genai:gemini-3.8-flash", trigger=("messages",8), keep=("messages",5))],
    checkpointer=InMemorySaver()
    )

questions = [
    "What is the capital of France?",
    "What is the capital of Germany?",
    "What is the capital of Italy?",
    "What is the capital of Spain?",
    "What is the capital of Portugal?"
    ]

for question in questions:
    response = agent.invoke({"messages": [HumanMessage(content=question)]}, config=config)
    print(f"end={response}")
    print(f"Message length ={len(response['messages'])}")

```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver
from langchain_core.messages import HumanMessage
from langchain.tools import tool

@tool
def search_hotels(
    location: str,
    check_in_date: str,
    check_out_date: str,
    num_guests: int,
) -> str:
    """
    Search for hotels in a given location and return a list of available hotels.
    """
    # For demonstration purposes, we'll return a static list of hotels.
    # In a real implementation, you would query a hotel database or API.
    return f"Available hotels in {location} from {check_in_date} to {check_out_date} for {num_guests} guests: Hotel A, Hotel B, Hotel C."

agent = create_agent(
    model="google_genai:gemini-3.8-flash",
    tools=[search_hotels],
    middleware=[SummarizationMiddleware(model="google_genai:gemini-3.8-flash", trigger=("tokens", 400), keep=("tokens", 250))],
    checkpointer=InMemorySaver()
    )

def count_tokens(messages):
    """
    Count the number of tokens in a list of messages.
    """
    total_chars = sum(len(str(message.content)) for message in messages)
    return total_chars // 4  # Approximate token count (1 token ≈ 4 characters)


cities = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]

for city in cities:
    response = agent.invoke({"messages": [HumanMessage(content=f"Find hotels in {city} from 2023-10-01 to 2023-10-05 for 2 guests.")]}, config=config)
    print(f"end={response}")
    print(f"Message length ={len(response['messages'])}")
    print(f"Token count ={count_tokens(response['messages'])}")

```
