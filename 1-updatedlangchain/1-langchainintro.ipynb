### Langchain Version v1

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')
os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
os.environ['GOOGLE_API_KEY'] = os.getenv('GOOGLE_API_KEY')
```

```python
from langchain.agents import create_agent

def get_weather(city:str)-> str:
    """this is weather function for a city"""
    # This function would contain logic to fetch weather information.
    # For demonstration purposes, we'll return a static response.
    return f"The current weather in {city} is sunny with a temperature of 25°C."   


agent = create_agent(
    model="google_genai:gemini-3.8-flash",
    tools=[get_weather],
    system_prompt="You are a helpful assistant that can answer questions about the world.",
)

agent
```

```python
messages = {"messages": [{"role": "user", "content": "What is the weather like today in newyork?"}]}
agent.invoke(messages)
```
